## 4. Why This Model Is Appropriate

A traffic-routing problem naturally maps to a **graph-search problem**. The system has a clearly defined initial state (source junction), a goal state (destination), actions (move along connected roads), and path costs (estimated travel time).

### State representation

A state is represented by a city node:

`State = current intersection`

### Actions

From a node, an action is to travel to any directly connected neighbouring node.

### Path cost

For a road:

\[
TravelTime = \frac{Distance}{SpeedLimit}\times 60 \times CongestionFactor
\]

where:

- Distance is in km.
- Speed limit is in km/h.
- 60 converts hours to minutes.
- Congestion factor is ≥ 1.0.

The total route cost is the sum of travel times across all road segments.

### Heuristic

For informed search, the notebook uses a lower-bound travel-time estimate:

\[
h(n)=\frac{StraightLineDistance(n,Goal)}{MaximumSpeed}\times60
\]

The implementation uses an even stricter lower bound: the shortest **free-flow travel time** from the current node to the goal. Free-flow time ignores congestion, while the actual modeled travel time includes a congestion factor of at least 1.0. Therefore the heuristic cannot overestimate the remaining traffic cost and is admissible for A*.

This makes **A\*** particularly suitable for the weighted routing problem because it combines the actual cost already travelled, \(g(n)\), with the estimated remaining cost, \(h(n)\):

\[
f(n)=g(n)+h(n)
\]


## 5. Search Algorithm Classification

| Category | Algorithm | Main idea |
|---|---|---|
| Uninformed | BFS | Expands nodes level by level |
| Uninformed | DFS | Explores one branch deeply before backtracking |
| Uninformed | DLS | DFS with a maximum depth |
| Uninformed | IDS | Repeated depth-limited searches |
| Uninformed | Bidirectional | Searches forward and backward |
| Informed | Greedy Best-First | Uses heuristic \(h(n)\) |
| Informed | A* | Uses \(g(n)+h(n)\) |
| Local/Optimization | Hill Climbing | Iteratively moves to a better neighbouring state |
| Local/Optimization | Genetic Algorithm | Evolves a population of candidate routes |

### Model-selection justification

For a **weighted traffic network**, the primary objective is not simply reaching the destination; it is reaching it with low travel time. Therefore, algorithms that explicitly consider path cost are especially relevant.

- BFS and DFS are useful baselines because they show how uninformed exploration behaves when traffic cost is ignored.
- DLS and IDS demonstrate depth-controlled search.
- Bidirectional search can reduce exploration when both directions can be searched efficiently.
- Greedy Best-First demonstrates the effect of relying mainly on estimated distance.
- A* is well matched to the problem because it combines accumulated travel time with an admissible estimate of remaining travel time.
- Hill Climbing demonstrates local optimization and its susceptibility to local choices.
- Genetic Algorithm demonstrates population-based optimization and is useful for discussing scalability to larger routing spaces.

The experiments below measure actual behaviour, so the final discussion is based on **observed results as well as algorithmic properties**.

```python
# ============================================================
# 🚦 SMART CITY TRAFFIC ROUTING
# GOOGLE COLAB IMPLEMENTATION
# ============================================================
#
# UNINFORMED SEARCH:
#   1. BFS
#   2. DFS
#   3. DLS
#   4. IDS
#   5. Bidirectional Search
#
# INFORMED SEARCH:
#   6. Greedy Best First Search
#   7. A* Search
#
# LOCAL / OPTIMIZATION SEARCH:
#   8. Hill Climbing
#   9. Genetic Algorithm
#
# BENCHMARK:
#   Dijkstra's algorithm is used internally as a reference
#   for minimum travel-time cost.
#
# ============================================================


# ============================================================
# 0. IMPORTS
# ============================================================

from collections import deque
import heapq
import math
import random
import time
import pandas as pd
import matplotlib.pyplot as plt
import networkx as nx

import ipywidgets as widgets
from IPython.display import display, clear_output


# ============================================================
# 1. CITY ROAD DATA
# ============================================================
# Each road is:
# (node1, node2, distance_km, speed_kmph, congestion_factor)
#
# congestion_factor = 1.0 means free-flow conditions.
# Values above 1.0 represent additional traffic delay.
# ============================================================

ROAD_DATA = [
    ("City Center", "A", 2.0, 40, 1.10),
    ("City Center", "B", 2.5, 50, 1.00),
    ("City Center", "C", 3.0, 40, 1.40),

    ("A", "D", 2.0, 40, 1.00),
    ("A", "E", 2.8, 45, 1.30),

    ("B", "E", 1.8, 35, 1.60),
    ("B", "F", 2.7, 50, 1.00),

    ("C", "F", 2.0, 40, 1.10),
    ("C", "G", 3.2, 45, 1.00),

    ("D", "E", 1.5, 30, 1.20),
    ("D", "H", 2.5, 40, 1.00),

    ("E", "F", 1.6, 35, 1.50),
    ("E", "H", 2.0, 40, 1.00),
    ("E", "I", 2.8, 50, 1.20),

    ("F", "G", 1.5, 45, 1.30),
    ("F", "I", 2.2, 40, 1.00),

    ("G", "I", 1.7, 45, 1.00),
    ("G", "J", 2.5, 40, 1.20),

    ("H", "I", 1.6, 35, 1.10),
    ("H", "K", 2.8, 50, 1.00),

    ("I", "J", 1.5, 45, 1.00),
    ("I", "K", 2.0, 40, 1.30),

    ("J", "K", 1.4, 35, 1.20),
    ("J", "Destination", 3.0, 50, 1.10),
    ("K", "Destination", 2.2, 40, 1.00),
]


# Approximate map coordinates for visualization and heuristic.
# They are not real geographic coordinates.
COORDS = {
    "City Center": (0, 5),
    "A": (2, 6),
    "B": (2, 4),
    "C": (2, 2),
    "D": (4, 7),
    "E": (4, 5),
    "F": (4, 3),
    "G": (4, 1),
    "H": (6, 7),
    "I": (6, 5),
    "J": (6, 3),
    "K": (8, 6),
    "Destination": (10, 5),
}


# ============================================================
# 2. BUILD GRAPH
# ============================================================

graph = {node: {} for node in COORDS}
road_info = {}

for u, v, distance, speed, congestion in ROAD_DATA:

    travel_time = (distance / speed) * 60 * congestion

    info = {
        "distance": distance,
        "speed": speed,
        "congestion": congestion,
        "time": travel_time
    }

    graph[u][v] = info
    graph[v][u] = info

    road_info[tuple(sorted((u, v)))] = info


# ============================================================
# 3. BASIC UTILITIES
# ============================================================

ALL_NODES = list(graph.keys())
DEFAULT_START = "City Center"
DEFAULT_GOAL = "Destination"

MAX_SPEED = max(info["speed"] for info in road_info.values())


def edge_info(u, v):
    """Return road information for an adjacent pair."""
    return graph[u][v]


def route_distance(path):
    if not path:
        return 0.0

    return sum(
        edge_info(path[i], path[i + 1])["distance"]
        for i in range(len(path) - 1)
    )


def route_cost(path):
    """Total traffic-adjusted travel time in minutes."""
    if not path:
        return 0.0

    return sum(
        edge_info(path[i], path[i + 1])["time"]
        for i in range(len(path) - 1)
    )


def is_valid_path(path, start, goal):
    if not path:
        return False

    if path[0] != start or path[-1] != goal:
        return False

    if len(path) != len(set(path)):
        return False

    for i in range(len(path) - 1):
        if path[i + 1] not in graph[path[i]]:
            return False

    return True


def safe_route_cost(path, start, goal):
    if not is_valid_path(path, start, goal):
        return float("inf")
    return route_cost(path)


def route_summary(path, start, goal):
    if not path:
        return {
            "valid": False,
            "distance": float("inf"),
            "cost": float("inf"),
            "nodes": 0
        }

    return {
        "valid": is_valid_path(path, start, goal),
        "distance": route_distance(path),
        "cost": route_cost(path),
        "nodes": len(path)
    }


# ============================================================
# 4. ADMISSIBLE HEURISTIC
# ============================================================
# For a rigorous A* experiment, the heuristic is computed as
# the shortest possible FREE-FLOW travel time from each node
# to the goal.
#
# Free-flow time ignores congestion:
#     distance / speed * 60
#
# Actual traffic time is always >= free-flow time because
# congestion_factor >= 1. Therefore this is an optimistic
# lower bound on the remaining traffic cost and is admissible.
# ============================================================

def free_flow_time(u, v):
    info = edge_info(u, v)
    return (info["distance"] / info["speed"]) * 60


def free_flow_heuristic_table(goal):
    distances = {
        node: float("inf")
        for node in ALL_NODES
    }

    distances[goal] = 0.0

    pq = [(0.0, goal)]

    while pq:

        current_cost, current = heapq.heappop(pq)

        if current_cost > distances[current] + 1e-12:
            continue

        for neighbor in graph[current]:

            new_cost = (
                current_cost
                + free_flow_time(current, neighbor)
            )

            if new_cost < distances[neighbor]:

                distances[neighbor] = new_cost

                heapq.heappush(
                    pq,
                    (new_cost, neighbor)
                )

    return distances


def heuristic(node, goal):
    """
    Admissible lower bound on remaining travel time.

    It is the shortest free-flow travel time from node to goal.
    Because congestion can only increase travel time, this
    heuristic never overestimates the actual remaining cost.
    """

    return free_flow_heuristic_table(goal)[node]


# ============================================================
# 5. BFS
# ============================================================

def bfs(start, goal):
    queue = deque([(start, [start])])
    visited = {start}
    expanded = []

    while queue:

        current, path = queue.popleft()
        expanded.append(current)

        if current == goal:
            return path, expanded

        for neighbor in graph[current]:

            if neighbor not in visited:
                visited.add(neighbor)
                queue.append((neighbor, path + [neighbor]))

    return None, expanded


# ============================================================
# 6. DFS
# ============================================================

def dfs(start, goal):
    stack = [(start, [start])]
    visited = set()
    expanded = []

    while stack:

        current, path = stack.pop()

        if current in visited:
            continue

        visited.add(current)
        expanded.append(current)

        if current == goal:
            return path, expanded

        # Reverse order gives deterministic behaviour.
        for neighbor in reversed(list(graph[current].keys())):

            if neighbor not in visited:
                stack.append((neighbor, path + [neighbor]))

    return None, expanded


# ============================================================
# 7. DEPTH LIMITED SEARCH
# ============================================================

def dls(start, goal, limit):
    expanded = []
    found_path = None

    def recursive_dls(current, path, depth):

        nonlocal found_path

        expanded.append(current)

        if current == goal:
            found_path = path
            return True

        if depth == limit:
            return False

        for neighbor in graph[current]:

            if neighbor not in path:

                if recursive_dls(
                    neighbor,
                    path + [neighbor],
                    depth + 1
                ):
                    return True

        return False

    recursive_dls(start, [start], 0)

    return found_path, expanded


# ============================================================
# 8. ITERATIVE DEEPENING SEARCH
# ============================================================

def ids(start, goal, max_depth=20):

    all_expanded = []

    for depth in range(max_depth + 1):

        path, expanded = dls(
            start,
            goal,
            depth
        )

        all_expanded.extend(expanded)

        if path:
            return path, all_expanded

    return None, all_expanded


# ============================================================
# 9. BIDIRECTIONAL SEARCH
# ============================================================

def bidirectional_search(start, goal):

    if start == goal:
        return [start], [start]

    forward_queue = deque([start])
    backward_queue = deque([goal])

    forward_parent = {start: None}
    backward_parent = {goal: None}

    expanded = []

    def expand_forward():

        if not forward_queue:
            return None

        current = forward_queue.popleft()
        expanded.append(current)

        for neighbor in graph[current]:

            if neighbor not in forward_parent:
                forward_parent[neighbor] = current
                forward_queue.append(neighbor)

                if neighbor in backward_parent:
                    return neighbor

        return None

    def expand_backward():

        if not backward_queue:
            return None

        current = backward_queue.popleft()
        expanded.append(current)

        for neighbor in graph[current]:

            if neighbor not in backward_parent:
                backward_parent[neighbor] = current
                backward_queue.append(neighbor)

                if neighbor in forward_parent:
                    return neighbor

        return None

    while forward_queue and backward_queue:

        meeting = expand_forward()

        if meeting is not None:
            break

        meeting = expand_backward()

        if meeting is not None:
            break

    else:
        return None, expanded

    # Reconstruct start -> meeting
    left = []
    current = meeting

    while current is not None:
        left.append(current)
        current = forward_parent[current]

    left.reverse()

    # Reconstruct meeting -> goal
    right = []
    current = backward_parent.get(meeting)

    while current is not None:
        right.append(current)
        current = backward_parent[current]

    return left + right, expanded


# ============================================================
# 10. GREEDY BEST-FIRST SEARCH
# ============================================================

def greedy_best_first(start, goal):

    counter = 0

    priority_queue = [
        (heuristic(start, goal), counter, start, [start])
    ]

    visited = set()
    expanded = []

    while priority_queue:

        h, _, current, path = heapq.heappop(priority_queue)

        if current in visited:
            continue

        visited.add(current)
        expanded.append(current)

        if current == goal:
            return path, expanded

        for neighbor in graph[current]:

            if neighbor not in visited:

                counter += 1

                heapq.heappush(
                    priority_queue,
                    (
                        heuristic(neighbor, goal),
                        counter,
                        neighbor,
                        path + [neighbor]
                    )
                )

    return None, expanded


# ============================================================
# 11. A* SEARCH
# ============================================================

def a_star(start, goal):

    counter = 0

    h_table = free_flow_heuristic_table(goal)

    priority_queue = [
        (
            h_table[start],
            0.0,
            counter,
            start,
            [start]
        )
    ]

    best_g = {start: 0.0}
    expanded = []

    while priority_queue:

        f, g, _, current, path = heapq.heappop(
            priority_queue
        )

        if g > best_g.get(current, float("inf")) + 1e-12:
            continue

        expanded.append(current)

        if current == goal:
            return path, expanded

        for neighbor, info in graph[current].items():

            new_g = g + info["time"]

            if new_g < best_g.get(
                neighbor,
                float("inf")
            ):

                best_g[neighbor] = new_g
                counter += 1

                new_f = (
                    new_g
                    + h_table[neighbor]
                )

                heapq.heappush(
                    priority_queue,
                    (
                        new_f,
                        new_g,
                        counter,
                        neighbor,
                        path + [neighbor]
                    )
                )

    return None, expanded


# ============================================================
# 12. DIJKSTRA BENCHMARK
# ============================================================

def dijkstra_benchmark(start, goal):

    counter = 0

    pq = [(0.0, counter, start, [start])]
    best_g = {start: 0.0}
    expanded = []

    while pq:

        g, _, current, path = heapq.heappop(pq)

        if g > best_g.get(current, float("inf")) + 1e-12:
            continue

        expanded.append(current)

        if current == goal:
            return path, expanded

        for neighbor, info in graph[current].items():

            new_g = g + info["time"]

            if new_g < best_g.get(
                neighbor,
                float("inf")
            ):

                best_g[neighbor] = new_g
                counter += 1

                heapq.heappush(
                    pq,
                    (
                        new_g,
                        counter,
                        neighbor,
                        path + [neighbor]
                    )
                )

    return None, expanded


# ============================================================
# 13. ALL SIMPLE PATHS
# ============================================================

def all_simple_paths(start, goal, max_nodes=None):

    if max_nodes is None:
        max_nodes = len(ALL_NODES)

    paths = []

    def backtrack(current, path):

        if len(path) > max_nodes:
            return

        if current == goal:
            paths.append(path.copy())
            return

        for neighbor in graph[current]:

            if neighbor not in path:
                backtrack(
                    neighbor,
                    path + [neighbor]
                )

    backtrack(start, [start])

    return paths


# ============================================================
# 14. HILL CLIMBING
# ============================================================

def hill_climbing(start, goal):

    current = start
    path = [start]
    expanded = []
    history = [path.copy()]

    while current != goal:

        expanded.append(current)

        candidates = [
            n for n in graph[current]
            if n not in path
        ]

        if not candidates:
            break

        # Prefer a lower heuristic value, but consider
        # the immediate traffic-adjusted road cost too.
        current_score = (
            heuristic(current, goal),
            route_cost(path)
        )

        scored = []

        for neighbor in candidates:

            candidate_score = (
                heuristic(neighbor, goal),
                edge_info(current, neighbor)["time"]
            )

            scored.append(
                (candidate_score, neighbor)
            )

        scored.sort(key=lambda x: x[0])

        best_score, best_neighbor = scored[0]

        # If the heuristic does not improve, stop.
        if best_score[0] >= current_score[0]:
            break

        current = best_neighbor
        path.append(current)
        history.append(path.copy())

    if current == goal:
        expanded.append(current)
        return path, expanded, history

    # A small repair: if pure hill climbing gets stuck,
    # report failure instead of inventing an invalid route.
    return None, expanded, history


# ============================================================
# 15. GENETIC ALGORITHM HELPERS
# ============================================================

def random_valid_route(start, goal, rng):

    for _ in range(100):

        current = start
        path = [start]

        while current != goal:

            candidates = [
                n for n in graph[current]
                if n not in path
            ]

            if not candidates:
                break

            weights = []

            for n in candidates:
                h = heuristic(n, goal)
                weights.append(1.0 / (h + 0.1))

            current = rng.choices(
                candidates,
                weights=weights,
                k=1
            )[0]

            path.append(current)

            if len(path) > len(ALL_NODES):
                break

        if current == goal and is_valid_path(
            path, start, goal
        ):
            return path

    return None


def route_fitness(route):
    cost = route_cost(route)
    return 1.0 / (1.0 + cost)


def crossover(parent1, parent2, start, goal, rng):

    common = [
        node for node in parent1[1:-1]
        if node in parent2[1:-1]
    ]

    if not common:
        return parent1.copy()

    meeting = rng.choice(common)

    i = parent1.index(meeting)
    j = parent2.index(meeting)

    child = parent1[:i] + parent2[j:]

    if is_valid_path(child, start, goal):
        return child

    return parent2.copy()


def mutate_route(route, start, goal, rng, mutation_rate):

    if rng.random() >= mutation_rate:
        return route.copy()

    new_route = random_valid_route(
        start,
        goal,
        rng
    )

    if new_route is None:
        return route.copy()

    return new_route


def genetic_algorithm(
    start,
    goal,
    population_size=30,
    generations=60,
    mutation_rate=0.20,
    seed=42
):

    rng = random.Random(seed)

    population = []
    attempts = 0
    max_attempts = population_size * 100

    while (
        len(population) < population_size
        and attempts < max_attempts
    ):

        attempts += 1

        route = random_valid_route(
            start,
            goal,
            rng
        )

        if route is not None:
            population.append(route)

    if not population:
        return None, [], [], 0

    best_route = min(
        population,
        key=route_cost
    ).copy()

    progress = []
    evaluations = 0

    for generation in range(generations):

        current_best = min(
            population,
            key=route_cost
        )

        evaluations += len(population)

        if route_cost(current_best) < route_cost(best_route):
            best_route = current_best.copy()

        progress.append(
            (
                generation + 1,
                route_cost(best_route)
            )
        )

        ranked = sorted(
            population,
            key=route_fitness,
            reverse=True
        )

        elite_count = max(
            2,
            population_size // 3
        )

        parents = ranked[:elite_count]

        new_population = [
            best_route.copy()
        ]

        while len(new_population) < population_size:

            parent1, parent2 = rng.sample(
                parents,
                2
            )

            child = crossover(
                parent1,
                parent2,
                start,
                goal,
                rng
            )

            child = mutate_route(
                child,
                start,
                goal,
                rng,
                mutation_rate
            )

            if is_valid_path(child, start, goal):
                new_population.append(child)

        population = new_population

    evaluations += len(population)

    return (
        best_route,
        progress,
        population,
        evaluations
    )


# ============================================================
# 16. ALGORITHM WRAPPER
# ============================================================

def run_algorithm(name, start, goal, dls_limit=8):

    begin = time.perf_counter()

    if name == "BFS":
        path, expanded = bfs(start, goal)

    elif name == "DFS":
        path, expanded = dfs(start, goal)

    elif name == "DLS":
        path, expanded = dls(
            start,
            goal,
            dls_limit
        )

    elif name == "IDS":
        path, expanded = ids(
            start,
            goal,
            max_depth=max(20, len(ALL_NODES))
        )

    elif name == "Bidirectional Search":
        path, expanded = bidirectional_search(
            start,
            goal
        )

    elif name == "Greedy Best-First":
        path, expanded = greedy_best_first(
            start,
            goal
        )

    elif name == "A*":
        path, expanded = a_star(
            start,
            goal
        )

    elif name == "Hill Climbing":
        path, expanded, history = hill_climbing(
            start,
            goal
        )

    elif name == "Genetic Algorithm":
        path, progress, _, evaluations = genetic_algorithm(
            start,
            goal
        )

        expanded = [
            f"Fitness evaluation {i + 1}"
            for i in range(evaluations)
        ]
        history = progress

    else:
        raise ValueError(
            f"Unknown algorithm: {name}"
        )

    elapsed_ms = (
        time.perf_counter() - begin
    ) * 1000

    valid = is_valid_path(
        path,
        start,
        goal
    )

    if valid:
        distance = route_distance(path)
        cost = route_cost(path)
        gap = None
    else:
        distance = float("inf")
        cost = float("inf")
        gap = None

    return {
        "algorithm": name,
        "path": path,
        "expanded": expanded,
        "expanded_count": len(set(expanded)),
        "distance": distance,
        "cost": cost,
        "runtime_ms": elapsed_ms,
        "valid": valid,
        "history": history if name in (
            "Hill Climbing",
            "Genetic Algorithm"
        ) else None,
        "gap": gap
    }


# ============================================================
# 17. BENCHMARK + COMPARISON
# ============================================================

ALGORITHM_NAMES = [
    "BFS",
    "DFS",
    "DLS",
    "IDS",
    "Bidirectional Search",
    "Greedy Best-First",
    "A*",
    "Hill Climbing",
    "Genetic Algorithm"
]


def run_benchmark(start, goal):

    benchmark_path, benchmark_expanded = (
        dijkstra_benchmark(
            start,
            goal
        )
    )

    benchmark_cost = (
        route_cost(benchmark_path)
        if benchmark_path
        else float("inf")
    )

    results = []

    for name in ALGORITHM_NAMES:

        result = run_algorithm(
            name,
            start,
            goal
        )

        if (
            result["valid"]
            and math.isfinite(benchmark_cost)
        ):
            result["gap"] = (
                (result["cost"] - benchmark_cost)
                / benchmark_cost
            ) * 100

        results.append(result)

    return (
        benchmark_path,
        benchmark_cost,
        benchmark_expanded,
        results
    )


def results_dataframe(
    results,
    benchmark_cost
):

    rows = []

    for result in results:

        rows.append({
            "Algorithm": result["algorithm"],
            "Valid": "Yes" if result["valid"] else "No",
            "Travel Time (min)": (
                round(result["cost"], 2)
                if result["valid"]
                else None
            ),
            "Distance (km)": (
                round(result["distance"], 2)
                if result["valid"]
                else None
            ),
            "Nodes Expanded": result["expanded_count"],
            "Runtime (ms)": round(
                result["runtime_ms"],
                3
            ),
            "Optimality Gap (%)": (
                round(result["gap"], 2)
                if result["gap"] is not None
                else None
            )
        })

    return pd.DataFrame(rows)


# ============================================================
# 18. ROUTE DISPLAY
# ============================================================

def print_route_result(
    result,
    start,
    goal,
    benchmark_cost=None
):

    print("=" * 90)
    print(
        f"                 {result['algorithm']}"
    )
    print("=" * 90)

    if not result["valid"]:

        print("\n❌ No valid route found.")
        print(
            f"Nodes Expanded : {result['expanded_count']}"
        )
        print(
            f"Runtime        : {result['runtime_ms']:.3f} ms"
        )
        return

    path = result["path"]

    print("\n✅ ROUTE FOUND\n")
    print(" → ".join(path))

    print(
        f"\nTravel Time       : {result['cost']:.2f} minutes"
    )

    print(
        f"Road Distance    : {result['distance']:.2f} km"
    )

    print(
        f"Nodes Expanded   : {result['expanded_count']}"
    )

    print(
        f"Execution Time   : {result['runtime_ms']:.3f} ms"
    )

    if (
        benchmark_cost is not None
        and math.isfinite(benchmark_cost)
    ):

        print(
            f"Optimality Gap   : {result['gap']:.2f}%"
        )

        if abs(result["gap"]) < 1e-9:
            print(
                "Benchmark Check  : Matches Dijkstra cost"
            )
        else:
            print(
                "Benchmark Check  : Different from Dijkstra cost"
            )

    print(
        "\nExpanded Order:"
    )

    print(
        " → ".join(result["expanded"])
    )


# ============================================================
# 19. ROAD TABLE
# ============================================================

def display_road_table():

    rows = []

    for u, v, distance, speed, congestion in ROAD_DATA:

        time_min = (
            distance / speed
        ) * 60 * congestion

        rows.append({
            "Road": f"{u} ↔ {v}",
            "Distance (km)": distance,
            "Speed (km/h)": speed,
            "Congestion Factor": congestion,
            "Travel Time (min)": round(
                time_min,
                2
            )
        })

    return pd.DataFrame(rows)


# ============================================================
# 20. CITY MAP VISUALIZATION
# ============================================================

def plot_city_map(
    start=None,
    goal=None,
    path=None,
    title="Smart City Traffic Network"
):

    G = nx.Graph()

    for node, coord in COORDS.items():
        G.add_node(
            node,
            pos=coord
        )

    for u, v, distance, speed, congestion in ROAD_DATA:
        G.add_edge(
            u,
            v,
            distance=distance,
            speed=speed,
            congestion=congestion,
            time=(distance / speed) * 60 * congestion
        )

    plt.figure(figsize=(13, 8))

    # Draw normal roads.
    nx.draw_networkx_edges(
        G,
        COORDS,
        width=2,
        alpha=0.45
    )

    # Highlight congested roads.
    congested_edges = [
        (u, v)
        for u, v, d in G.edges(data=True)
        if d["congestion"] >= 1.3
    ]

    nx.draw_networkx_edges(
        G,
        COORDS,
        edgelist=congested_edges,
        width=4,
        alpha=0.35,
        edge_color="red"
    )

    nx.draw_networkx_nodes(
        G,
        COORDS,
        node_size=900,
        node_color="lightblue",
        edgecolors="black"
    )

    if start in G:

        nx.draw_networkx_nodes(
            G,
            COORDS,
            nodelist=[start],
            node_size=1100,
            node_color="limegreen",
            edgecolors="black"
        )

    if goal in G:

        nx.draw_networkx_nodes(
            G,
            COORDS,
            nodelist=[goal],
            node_size=1100,
            node_color="orange",
            edgecolors="black"
        )

    nx.draw_networkx_labels(
        G,
        COORDS,
        font_size=9,
        font_weight="bold"
    )

    if path and len(path) >= 2:

        route_edges = list(
            zip(
                path[:-1],
                path[1:]
            )
        )

        nx.draw_networkx_edges(
            G,
            COORDS,
            edgelist=route_edges,
            width=5,
            edge_color="blue"
        )

    edge_labels = {}

    for u, v, d in G.edges(data=True):

        edge_labels[(u, v)] = (
            f"{d['time']:.1f}m | C={d['congestion']:.1f}"
        )

    nx.draw_networkx_edge_labels(
        G,
        COORDS,
        edge_labels=edge_labels,
        font_size=7
    )

    plt.title(
        title,
        fontsize=16,
        fontweight="bold"
    )

    plt.axis("off")
    plt.show()


# ============================================================
# 21. COMPARISON VISUALIZATION
# ============================================================

def plot_comparison(df):

    valid_df = df.dropna(
        subset=["Travel Time (min)"]
    )

    if valid_df.empty:
        print("No valid results to plot.")
        return

    plt.figure(figsize=(12, 6))

    plt.bar(
        valid_df["Algorithm"],
        valid_df["Travel Time (min)"]
    )

    plt.ylabel("Travel Time (minutes)")
    plt.xlabel("Algorithm")
    plt.title(
        "Travel-Time Comparison"
    )

    plt.xticks(
        rotation=35,
        ha="right"
    )

    plt.tight_layout()
    plt.show()


def plot_nodes_expanded(df):

    plt.figure(figsize=(12, 6))

    plt.bar(
        df["Algorithm"],
        df["Nodes Expanded"]
    )

    plt.ylabel("Unique Nodes Expanded")
    plt.xlabel("Algorithm")
    plt.title(
        "Search Effort Comparison"
    )

    plt.xticks(
        rotation=35,
        ha="right"
    )

    plt.tight_layout()
    plt.show()


def plot_optimality_gap(df):

    valid_df = df.dropna(
        subset=["Optimality Gap (%)"]
    )

    if valid_df.empty:
        return

    plt.figure(figsize=(12, 6))

    plt.bar(
        valid_df["Algorithm"],
        valid_df["Optimality Gap (%)"]
    )

    plt.axhline(
        0,
        linewidth=1
    )

    plt.ylabel("Optimality Gap (%)")
    plt.xlabel("Algorithm")
    plt.title(
        "Optimality Gap Relative to Dijkstra"
    )

    plt.xticks(
        rotation=35,
        ha="right"
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 22. TRAFFIC CONDITION SUMMARY
# ============================================================

def traffic_summary():

    congestion_values = [
        c for _, _, _, _, c in ROAD_DATA
    ]

    print("SMART CITY TRAFFIC DATASET")
    print("=" * 60)
    print(
        f"Number of locations : {len(ALL_NODES)}"
    )
    print(
        f"Number of road links: {len(ROAD_DATA)}"
    )
    print(
        f"Average congestion  : "
        f"{sum(congestion_values)/len(congestion_values):.2f}"
    )
    print(
        f"Highest congestion  : "
        f"{max(congestion_values):.2f}"
    )
    print(
        f"Maximum speed       : "
        f"{MAX_SPEED:.0f} km/h"
    )

    print(
        "\nCongestion interpretation:"
    )
    print(
        "1.00 - free-flow baseline"
    )
    print(
        "1.10 - light additional delay"
    )
    print(
        "1.20 - moderate delay"
    )
    print(
        "1.30+ - heavier congestion"
    )


# ============================================================
# 23. EXPERIMENT FUNCTION
# ============================================================

def complete_experiment(
    start=DEFAULT_START,
    goal=DEFAULT_GOAL
):

    print(
        f"\nSOURCE      : {start}"
    )

    print(
        f"DESTINATION : {goal}\n"
    )

    (
        benchmark_path,
        benchmark_cost,
        benchmark_expanded,
        results
    ) = run_benchmark(
        start,
        goal
    )

    print(
        "Dijkstra Benchmark"
    )

    print(
        " → ".join(benchmark_path)
    )

    print(
        f"Benchmark Travel Time : "
        f"{benchmark_cost:.2f} minutes"
    )

    print(
        f"Benchmark Nodes       : "
        f"{len(set(benchmark_expanded))}"
    )

    df = results_dataframe(
        results,
        benchmark_cost
    )

    display(df)

    return (
        benchmark_path,
        benchmark_cost,
        results,
        df
    )


# ============================================================
# 24. INTERACTIVE GOOGLE COLAB GUI
# ============================================================

output = widgets.Output()

title = widgets.HTML(
    value="""
    <div style="
        text-align:center;
        font-size:28px;
        font-weight:bold;
        padding:12px;
    ">
        🚦 SMART CITY TRAFFIC ROUTING
    </div>

    <div style="
        text-align:center;
        color:#666;
        font-size:15px;
        margin-bottom:12px;
    ">
        Artificial Intelligence Search Algorithms
    </div>
    """
)


start_dropdown = widgets.Dropdown(
    options=ALL_NODES,
    value=DEFAULT_START,
    description="Source:",
    layout=widgets.Layout(width="320px")
)


goal_dropdown = widgets.Dropdown(
    options=ALL_NODES,
    value=DEFAULT_GOAL,
    description="Goal:",
    layout=widgets.Layout(width="320px")
)


depth_input = widgets.IntText(
    value=8,
    min=0,
    max=30,
    step=1,
    description="DLS limit:",
    layout=widgets.Layout(width="200px")
)


def make_button(text, style="primary"):
    return widgets.Button(
        description=text,
        button_style=style,
        layout=widgets.Layout(
            width="300px",
            height="40px",
            margin="4px"
        )
    )


bfs_button = make_button(
    "1. Breadth First Search (BFS)"
)

dfs_button = make_button(
    "2. Depth First Search (DFS)"
)

dls_button = make_button(
    "3. Depth Limited Search (DLS)"
)

ids_button = make_button(
    "4. Iterative Deepening Search (IDS)"
)

bidirectional_button = make_button(
    "5. Bidirectional Search"
)

greedy_button = make_button(
    "6. Greedy Best-First Search"
)

astar_button = make_button(
    "7. A* Search"
)

hill_button = make_button(
    "8. Hill Climbing"
)

genetic_button = make_button(
    "9. Genetic Algorithm"
)

compare_button = make_button(
    "10. Compare All Algorithms"
)

map_button = make_button(
    "11. Display City Map"
)

data_button = make_button(
    "12. Display Road Data"
)

exit_button = make_button(
    "13. Exit",
    "danger"
)


# ============================================================
# BUTTON CALLBACK
# ============================================================

def execute_algorithm(name):

    start = start_dropdown.value
    goal = goal_dropdown.value

    if start == goal:

        with output:

            clear_output(wait=True)

            print(
                "⚠️ Source and destination must be different."
            )

        return

    result = run_algorithm(
        name,
        start,
        goal,
        dls_limit=depth_input.value
    )

    benchmark_path, benchmark_expanded = (
        dijkstra_benchmark(
            start,
            goal
        )
    )

    benchmark_cost = route_cost(
        benchmark_path
    )

    if result["valid"]:

        result["gap"] = (
            (result["cost"] - benchmark_cost)
            / benchmark_cost
        ) * 100

    with output:

        clear_output(wait=True)

        print_route_result(
            result,
            start,
            goal,
            benchmark_cost
        )

        if result["valid"]:

            print(
                "\n📍 ROUTE MAP\n"
            )

            plot_city_map(
                start,
                goal,
                result["path"],
                f"{name}: {start} → {goal}"
            )


def on_bfs(_):
    execute_algorithm("BFS")


def on_dfs(_):
    execute_algorithm("DFS")


def on_dls(_):
    execute_algorithm("DLS")


def on_ids(_):
    execute_algorithm("IDS")


def on_bidirectional(_):
    execute_algorithm("Bidirectional Search")


def on_greedy(_):
    execute_algorithm("Greedy Best-First")


def on_astar(_):
    execute_algorithm("A*")


def on_hill(_):
    execute_algorithm("Hill Climbing")


def on_genetic(_):

    start = start_dropdown.value
    goal = goal_dropdown.value

    if start == goal:

        with output:

            clear_output(wait=True)

            print(
                "⚠️ Source and destination must be different."
            )

        return

    result = run_algorithm(
        "Genetic Algorithm",
        start,
        goal
    )

    benchmark_path, _ = dijkstra_benchmark(
        start,
        goal
    )

    benchmark_cost = route_cost(
        benchmark_path
    )

    if result["valid"]:

        result["gap"] = (
            (result["cost"] - benchmark_cost)
            / benchmark_cost
        ) * 100

    with output:

        clear_output(wait=True)

        print_route_result(
            result,
            start,
            goal,
            benchmark_cost
        )

        if result["valid"]:

            progress = result["history"]

            if progress:

                generations = [
                    p[0] for p in progress
                ]

                costs = [
                    p[1] for p in progress
                ]

                plt.figure(figsize=(10, 5))

                plt.plot(
                    generations,
                    costs,
                    marker="o",
                    markersize=3
                )

                plt.xlabel("Generation")
                plt.ylabel("Best Travel Time (min)")
                plt.title(
                    "Genetic Algorithm Convergence"
                )

                plt.grid(alpha=0.25)
                plt.show()

            plot_city_map(
                start,
                goal,
                result["path"],
                "Genetic Algorithm Route"
            )


def on_compare(_):

    start = start_dropdown.value
    goal = goal_dropdown.value

    if start == goal:

        with output:

            clear_output(wait=True)

            print(
                "⚠️ Source and destination must be different."
            )

        return

    with output:

        clear_output(wait=True)

        (
            benchmark_path,
            benchmark_cost,
            results,
            df
        ) = complete_experiment(
            start,
            goal
        )

        print(
            "\n📊 TRAVEL TIME COMPARISON"
        )

        plot_comparison(df)

        print(
            "\n📊 SEARCH EFFORT COMPARISON"
        )

        plot_nodes_expanded(df)

        print(
            "\n📊 OPTIMALITY GAP"
        )

        plot_optimality_gap(df)

        print(
            "\n🗺️ BENCHMARK ROUTE"
        )

        print(
            " → ".join(benchmark_path)
        )

        plot_city_map(
            start,
            goal,
            benchmark_path,
            "Dijkstra Benchmark Route"
        )


def on_map(_):

    with output:

        clear_output(wait=True)

        plot_city_map(
            start_dropdown.value,
            goal_dropdown.value,
            None,
            "Smart City Traffic Network"
        )


def on_data(_):

    with output:

        clear_output(wait=True)

        traffic_summary()

        print(
            "\nROAD-LEVEL DATA"
        )

        display(
            display_road_table()
        )


def on_exit(_):

    with output:

        clear_output(wait=True)

        print(
            "Session closed. You can rerun the notebook "
            "or select another option."
        )


# ============================================================
# CONNECT CALLBACKS
# ============================================================

bfs_button.on_click(on_bfs)
dfs_button.on_click(on_dfs)
dls_button.on_click(on_dls)
ids_button.on_click(on_ids)
bidirectional_button.on_click(on_bidirectional)

greedy_button.on_click(on_greedy)
astar_button.on_click(on_astar)

hill_button.on_click(on_hill)
genetic_button.on_click(on_genetic)

compare_button.on_click(on_compare)
map_button.on_click(on_map)
data_button.on_click(on_data)
exit_button.on_click(on_exit)


# ============================================================
# GUI LAYOUT
# ============================================================

location_section = widgets.VBox(
    [
        widgets.HTML(
            "<h3 style='text-align:center;'>📍 ROUTE SELECTION</h3>"
        ),
        widgets.HBox(
            [
                start_dropdown,
                goal_dropdown
            ],
            layout=widgets.Layout(
                justify_content="center"
            )
        ),
        widgets.HBox(
            [
                depth_input
            ],
            layout=widgets.Layout(
                justify_content="center"
            )
        )
    ],
    layout=widgets.Layout(
        border="2px solid #93c5fd",
        padding="12px",
        margin="8px"
    )
)


uninformed_section = widgets.VBox(
    [
        widgets.HTML(
            """
            <div style='font-size:19px;font-weight:bold;'>
            🔍 UNINFORMED SEARCH
            </div>
            <div style='color:#555;font-size:13px;'>
            Search without heuristic information
            </div>
            """
        ),

        widgets.HBox(
            [bfs_button, dfs_button],
            layout=widgets.Layout(
                justify_content="center"
            )
        ),

        widgets.HBox(
            [dls_button, ids_button],
            layout=widgets.Layout(
                justify_content="center"
            )
        ),

        widgets.HBox(
            [bidirectional_button],
            layout=widgets.Layout(
                justify_content="center"
            )
        )
    ],
    layout=widgets.Layout(
        border="2px solid #86efac",
        padding="12px",
        margin="8px"
    )
)


informed_section = widgets.VBox(
    [
        widgets.HTML(
            """
            <div style='font-size:19px;font-weight:bold;'>
            🧠 INFORMED SEARCH
            </div>
            <div style='color:#555;font-size:13px;'>
            Search using heuristic information
            </div>
            """
        ),

        widgets.HBox(
            [greedy_button, astar_button],
            layout=widgets.Layout(
                justify_content="center"
            )
        )
    ],
    layout=widgets.Layout(
        border="2px solid #93c5fd",
        padding="12px",
        margin="8px"
    )
)


local_section = widgets.VBox(
    [
        widgets.HTML(
            """
            <div style='font-size:19px;font-weight:bold;'>
            ⛰️ LOCAL / OPTIMIZATION SEARCH
            </div>
            <div style='color:#555;font-size:13px;'>
            Local and population-based optimization
            </div>
            """
        ),

        widgets.HBox(
            [hill_button, genetic_button],
            layout=widgets.Layout(
                justify_content="center"
            )
        )
    ],
    layout=widgets.Layout(
        border="2px solid #fcd34d",
        padding="12px",
        margin="8px"
    )
)


other_section = widgets.VBox(
    [
        widgets.HTML(
            """
            <div style='font-size:19px;font-weight:bold;'>
            ⚙️ OTHER OPTIONS
            </div>
            """
        ),

        widgets.HBox(
            [compare_button, map_button],
            layout=widgets.Layout(
                justify_content="center"
            )
        ),

        widgets.HBox(
            [data_button, exit_button],
            layout=widgets.Layout(
                justify_content="center"
            )
        )
    ],
    layout=widgets.Layout(
        border="1px solid #cccccc",
        padding="12px",
        margin="8px"
    )
)


display(
    widgets.VBox(
        [
            title,
            location_section,
            uninformed_section,
            informed_section,
            local_section,
            other_section,
            widgets.HTML(
                "<h3 style='text-align:center;'>📊 RESULT</h3>"
            ),
            output
        ],
        layout=widgets.Layout(
            width="100%"
        )
    )
)

print(
    "✅ Smart City Traffic Routing System loaded successfully."
)
print(
    "Select a source and destination, then run an algorithm."
)
print(
    "Use 'Compare All Algorithms' for the experimental analysis."
)
```
