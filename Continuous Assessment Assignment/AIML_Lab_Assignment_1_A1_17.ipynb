**8. Model Selection and Justification:-**

**Model Selection:**

The selected AI technique is a Simple Reflex Agent.

**Justification:**

Simple Reflex Agent would be suitable since the agent's action can be deduced based on the current perception of the environment by the agent. The hallway is divided into various places where each place will have any of the four possible conditions. There is an action associated with each condition. Therefore, the agent can use the condition-action rule without using history.

0 – Clean & Dry: Do nothing

1 – Dusty: Vacuuming + Brushing

2 – Wet: Water sucking

3 – Wet & Dusty: Vacuuming + Brushing + Water sucking

**Comparison with Other Agent:-**

Simple Reflex Agent appears to be best suited for the current simulation because it makes action selection based on the environmental state. A Model-Based Agent will not be required for this system as there will be no need for any kind of environment representation inside the agent. The Goal-Based Agent may become too complicated. A Learning Agent may be needed in the future to make the robot learn from its sensors while performing its duties.

**11. Code:-**

```python
import random

# FLOOR CONDITION CONSTANTS

CLEAN = 0
DUSTY = 1
WET = 2
WET_DUSTY = 3

# ENVIRONMENT

class Environment:

    def __init__(self, number_of_locations=6):

        self.locations = {}

        # Create multiple corridor locations
        for i in range(number_of_locations):
            location = chr(65 + i)   # A, B, C, D, E, F...
            self.locations[location] = random.randint(0, 3)

    def display(self):
        print(self.locations)

# SMART WET AND DRY VACUUM AGENT

class SmartWetDryVacuumAgent:

    def __init__(self, environment):

        self.environment = environment

        # Performance metrics
        self.cleaning_reward = 0
        self.movement_cost = 0
        self.cleaning_operations = 0
        self.movement_count = 0

        # Condition counters
        self.initial_clean = 0
        self.initial_dusty = 0
        self.initial_wet = 0
        self.initial_wet_dusty = 0

        self.cleaned_locations = 0

        # Display initial environment
        print("\n========== INITIAL ENVIRONMENT ==========")
        self.environment.display()

        # Count initial conditions
        self.count_conditions()

        # Start cleaning
        self.clean_corridor()

        # Display final results
        self.display_results()

    # SENSOR / PERCEPTION

    def sense_condition(self, location):

        return self.environment.locations[location]

    # COUNT INITIAL CONDITIONS

    def count_conditions(self):

        for condition in self.environment.locations.values():

            if condition == CLEAN:
                self.initial_clean += 1

            elif condition == DUSTY:
                self.initial_dusty += 1

            elif condition == WET:
                self.initial_wet += 1

            elif condition == WET_DUSTY:
                self.initial_wet_dusty += 1

    # CLEAN CURRENT LOCATION

    def clean_location(self, location):

        condition = self.sense_condition(location)

        print("\nChecking Location:", location)

        # CLEAN AND DRY
        if condition == CLEAN:

            print("Condition: Clean and Dry")
            print("Action: No cleaning required.")

        # DUSTY
        elif condition == DUSTY:

            print("Condition: Dusty")
            print("Action: Activating Dry Vacuum and Brush.")

            self.environment.locations[location] = CLEAN

            self.cleaning_reward += 1
            self.cleaning_operations += 1
            self.cleaned_locations += 1

            print("Result: Dust removed successfully.")

        # WET
        elif condition == WET:

            print("Condition: Wet")
            print("Action: Activating Water Suction System.")

            self.environment.locations[location] = CLEAN

            self.cleaning_reward += 1
            self.cleaning_operations += 1
            self.cleaned_locations += 1

            print("Result: Water removed successfully.")

        # WET AND DUSTY
        elif condition == WET_DUSTY:

            print("Condition: Wet and Dusty")
            print("Action: Activating Vacuum, Brush and Water Suction.")

            self.environment.locations[location] = CLEAN

            self.cleaning_reward += 2
            self.cleaning_operations += 1
            self.cleaned_locations += 1

            print("Result: Dust and water removed successfully.")

    # MOVEMENT

    def move_to_next_location(self, current, next_location):

        print("\nMoving from Location", current,
              "to Location", next_location)

        self.movement_count += 1
        self.movement_cost += 1

    # CLEAN COMPLETE CORRIDOR

    def clean_corridor(self):

        locations = list(self.environment.locations.keys())

        # Random starting location
        start_index = random.randint(0, len(locations) - 1)

        # Rotate list so cleaning starts at random location
        locations = (
            locations[start_index:] +
            locations[:start_index]
        )

        print("\nVacuum randomly starts at Location:",
              locations[0])

        # Visit every location
        for i, location in enumerate(locations):

            self.clean_location(location)

            # Move to next location
            if i < len(locations) - 1:

                self.move_to_next_location(
                    location,
                    locations[i + 1]
                )

    # PERFORMANCE CALCULATION

    def calculate_score(self):

        return self.cleaning_reward - self.movement_cost

    def calculate_cleaning_efficiency(self):

        total_locations = len(self.environment.locations)

        if total_locations == 0:
            return 0

        return (
            self.cleaned_locations /
            total_locations
        ) * 100

    # FINAL REPORT

    def display_results(self):

        final_score = self.calculate_score()
        efficiency = self.calculate_cleaning_efficiency()

        print("\n==========================================")
        print("             FINAL RESULTS")
        print("==========================================")

        print("\nFinal Environment:")
        self.environment.display()

        print("\nInitial Condition Distribution:")
        print("Clean and Dry :", self.initial_clean)
        print("Dusty         :", self.initial_dusty)
        print("Wet           :", self.initial_wet)
        print("Wet and Dusty :", self.initial_wet_dusty)

        print("\nPerformance Metrics:")
        print("Total Locations    :", len(self.environment.locations))
        print("Locations Cleaned  :", self.cleaned_locations)
        print("Cleaning Operations:", self.cleaning_operations)
        print("Total Movements    :", self.movement_count)
        print("Cleaning Reward    :", self.cleaning_reward)
        print("Movement Cost      :", self.movement_cost)
        print("Final Performance  :", final_score)
        print("Cleaning Efficiency: {:.2f}%".format(efficiency))

        print("\n==========================================")


# MAIN PROGRAM

if __name__ == "__main__":

    # Create environment with 6 corridor locations
    environment = Environment(number_of_locations=6)

    # Create Smart Wet and Dry Vacuum Agent
    vacuum = SmartWetDryVacuumAgent(environment)
```
